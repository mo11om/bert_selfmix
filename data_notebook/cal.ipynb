```python
import pandas as pd

# This assumes you have a CSV file at 'data/split/test.csv'
# with 'noise', 'text', and 'tag' columns.
file_path = "data/split/test.csv"
output_path = "data/clean/test.csv"

try:
    df = pd.read_csv(file_path)

    # Filter the DataFrame to get rows where the 'noise' column is 0
    noise_is_zero_df = df[df['noise'] == 0]

    # Select only the 'text' and 'tag' columns
    filtered_columns_df = noise_is_zero_df[[ 'tag','text']]
    print(filtered_columns_df.head())
    # Save the new DataFrame to a CSV file, without the index column
    filtered_columns_df.to_csv(output_path, index=False, header=False)

    print(f"Successfully created '{output_path}' with 'text' and 'tag' columns where noise was 0.")

except FileNotFoundError:
    print(f"Error: The file at '{file_path}' was not found.")
    print("Please make sure the file path is correct and the file exists.")
except KeyError as e:
    print(f"Error: A required column was not found in the file: {e}")
    print("Please make sure your CSV file contains the 'noise', 'text', and 'tag' columns.")
```
